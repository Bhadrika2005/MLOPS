```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer


df = pd.read_csv("/content/drive/MyDrive/hiring.csv")


word_to_num = {
    'zero': 0,
    'one': 1,
    'two': 2,
    'three': 3,
    'four': 4,
    'five': 5,
    'six': 6,
    'seven': 7,
    'eight': 8,
    'nine': 9,
    'ten': 10,
    'eleven': 11,
    'twelve': 12
}


df['experience'] = df['experience'].fillna('zero')


df['experience'] = df['experience'].map(word_to_num)


imputer = SimpleImputer(strategy='median')
df[['test_score(out of 10)']] = imputer.fit_transform(
    df[['test_score(out of 10)']]
)


X = df[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']]
y = df['salary($)']


model = LinearRegression()
model.fit(X, y)


candidate1 = pd.DataFrame({
    'experience': [2],
    'test_score(out of 10)': [9],
    'interview_score(out of 10)': [6]
})


candidate2 = pd.DataFrame({
    'experience': [12],
    'test_score(out of 10)': [10],
    'interview_score(out of 10)': [10]
})


salary1 = model.predict(candidate1)
salary2 = model.predict(candidate2)


print("Predicted salary for candidate 1:")
print(salary1[0])

print("\nPredicted salary for candidate 2:")
print(salary2[0])
```
