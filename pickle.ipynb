```python

import pandas as pd
from sklearn.linear_model import LinearRegression
import pickle


df = pd.read_csv('/content/drive/MyDrive/forensic_scene_regression_dataset1.csv')


print(df.head())


X = df.iloc[:, :-1]


y = df.iloc[:, -1]


model = LinearRegression()


model.fit(X, y)


with open('crimevision_ai_model.pkl', 'wb') as file:
    pickle.dump(model, file)

print("Pickle file created successfully!")
```
