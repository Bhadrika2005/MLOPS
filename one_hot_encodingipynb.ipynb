```python

import pandas as pd
from sklearn.linear_model import LinearRegression
import pickle


df = pd.read_csv('/content/drive/MyDrive/forensic_scene_regression_dataset1.csv')


print("Original Dataset:")
print(df.head())


incident_list = ['burglary', 'fire', 'vandalism', 'assault']

df['incident_type'] = [
    incident_list[i % 4] for i in range(len(df))
]


print("\nDataset with Incident Type:")
print(df.head())


encoded_df = pd.get_dummies(df, columns=['incident_type'])


encoded_df = encoded_df.astype(int)


print("\nEncoded Dataset:")
print(encoded_df.head())


X = encoded_df.drop('suspicious_score', axis=1)
y = encoded_df['suspicious_score']


model = LinearRegression()


model.fit(X, y)


with open('crimevision_ai_model.pkl', 'wb') as file:
    pickle.dump(model, file)

print("\nModel trained successfully!")
print("Pickle file 'crimevision_ai_model.pkl' created successfully.")
```
